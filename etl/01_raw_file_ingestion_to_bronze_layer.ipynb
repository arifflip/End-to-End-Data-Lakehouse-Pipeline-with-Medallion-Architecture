```python
# define catalgo and schema for the database
project_catalog='data_project'
project_schema='bronze'
table_name="raw_scraped_result"
volume_path='/Volumes/data_project/bronze/raw_scraped_data'

spark.sql(f"USE CATALOG {project_catalog}")
```

```python
# create bronze tabel
spark.sql(f""" CREATE TABLE IF NOT EXISTS {project_schema}.{table_name} """)
```

```python
# ingest csv file to bronze layer
spark.sql(f"""
    COPY INTO {project_schema}.{table_name}
    FROM '{volume_path}'
    FILEFORMAT = CSV
    PATTERN = '*.csv'
    FORMAT_OPTIONS (
        'header' = 'true',
        'delimiter' = '\\t',
        'inferSchema' = 'true'
    )
    COPY_OPTIONS (
        'mergeSchema' = 'true'
    )
""")
```
